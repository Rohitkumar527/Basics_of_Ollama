GENERATE USING OLLAMA LIBRARY

```python
import ollama
response = ollama.generate(model = "qwen3:4b",prompt = "explain black holes in one sentence")

print(response.response)
```

GENERATE USING REST API

```python
import requests
import json

url = "http://localhost:11434/api/generate"

payload = {
    "model":"qwen3:4b","prompt":"explain black holes in simple"}

response = requests.post(url,json = payload)
```

```python
response
```

```python
for i in response.iter_lines():
    print(i)
```

```python
output = ""
for i in response.iter_lines():
    if i:
        data = json.loads(i.decode("utf-8"))
        if "response" in data:
            output +=data["response"]
        if data.get("done"):
            break


print(output)            
```
